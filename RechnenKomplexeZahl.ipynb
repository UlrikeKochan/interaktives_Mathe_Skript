## Addition, Subtraktion, Multiplikation und Division

### Addition und Subtraktion von komplexen Zahlen

**Addition und Subtraktion:**

- Kombiniere gleichartige Terme (reell mit reell, imaginär mit imaginär):

$$
(a+bi)+(c+di)
=
(a+c)+(b+d)i
$$

$$
(a+bi)-(c+di)
=
(a-c)+(b-d)i
$$

**Beispiel:**

$$
(3+2i)+(1+4i)
=
4+6i
$$

$$
(5-3i)-(2+i)
=
3-4i
$$

### Graphische Darstellung der Addition

Die Addition komplexer Zahlen kann in der Gauss-Ebene als Vektoraddition dargestellt werden.

Für

$$
z_1=3+2i
$$

und

$$
z_2=1+4i
$$

ergibt sich

$$
z_1+z_2
=
4+6i.
$$

```python
import matplotlib.pyplot as plt

z1 = (3, 2)
z2 = (1, 4)
z_sum = (4, 6)

fig, ax = plt.subplots(figsize=(7, 7))

# Achsen
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

# z1
ax.arrow(
    0, 0, z1[0], z1[1],
    length_includes_head=True,
    head_width=0.15
)

# z2 verschoben an die Spitze von z1
ax.arrow(
    z1[0], z1[1], z2[0], z2[1],
    length_includes_head=True,
    head_width=0.15
)

# Summe
ax.arrow(
    0, 0, z_sum[0], z_sum[1],
    length_includes_head=True,
    head_width=0.15
)

# z2 zusätzlich vom Ursprung gestrichelt
ax.plot([0, z2[0]], [0, z2[1]], linestyle="--")

ax.text(2.3, 1.1, r"$3+2i$", fontsize=12)
ax.text(3.2, 4.3, r"$1+4i$", fontsize=12)
ax.text(2.0, 4.8, r"$4+6i$", fontsize=12)

ax.set_xlim(-1, 5)
ax.set_ylim(-1, 7)
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Multiplikation von komplexen Zahlen

**Multiplikation:**

Nutze das Distributivgesetz und die Tatsache, dass

$$
i^2=-1.
$$

Dann gilt:

$$
(a+bi)(c+di)
=
(ac-bd)+(ad+bc)i.
$$

**Beispiel:**

$$
(2+3i)(4-i)
=
(2\cdot4-3\cdot(-1))
+
(2\cdot(-1)+3\cdot4)i
$$

$$
=
(8+3)+(-2+12)i
=
11+10i.
$$

### Graphische Darstellung der Multiplikation

Bei der Multiplikation zweier komplexer Zahlen werden

- die Beträge (Längen) multipliziert,
- die Winkel (Argumente) addiert.

Beispiel:

$$
(2+3i)(4-i)
=
11+10i.
$$

Grafisch entspricht dies einer Skalierung und Drehung des Vektors.

### Winkelrichtung für komplexe Zahlen

**Winkelmessung:**

Der Winkel $\theta$ eines Vektors im komplexen Zahlenbereich kann auf zwei Arten gemessen werden:

- **Positiver Winkel:** Gemessen gegen den Uhrzeigersinn von der positiven x-Achse.
- **Negativer Winkel:** Gemessen im Uhrzeigersinn von der positiven x-Achse.

Diese beiden Richtungen repräsentieren denselben Vektor, da beide Winkel dieselbe Orientierung im Raum beschreiben.

**Beispiel:**

Der Vektor

$$
4-i
$$

kann mit einem Winkel von etwa

$$
\theta_2=346^\circ
$$

gegen den Uhrzeigersinn oder mit

$$
\theta_2=-14^\circ
$$

im Uhrzeigersinn beschrieben werden.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Arc

fig, ax = plt.subplots(figsize=(7, 5))

z = np.array([4, -1])

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

ax.arrow(
    0, 0, z[0], z[1],
    length_includes_head=True,
    head_width=0.12
)

ax.text(3.0, -1.25, r"$4-i$", fontsize=12)

# kleiner negativer Winkel
arc_neg = Arc(
    (0, 0),
    2.0, 2.0,
    angle=0,
    theta1=-14,
    theta2=0,
    linewidth=1.5
)
ax.add_patch(arc_neg)

ax.text(1.1, -0.3, r"$-14^\circ$", fontsize=11)

# Hinweis auf äquivalenten positiven Winkel
ax.text(
    -0.8, 1.5,
    r"$346^\circ = -14^\circ + 360^\circ$",
    fontsize=11
)

ax.set_xlim(-1.5, 5)
ax.set_ylim(-2, 2.5)
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Multiplikation: Winkeladdition

Für

$$
z_1=2+3i
$$

und

$$
z_2=4-i
$$

gelten näherungsweise die Winkel

$$
\theta_1=56^\circ
$$

und

$$
\theta_2=-14^\circ.
$$

Damit gilt

$$
\theta_1+\theta_2
=
42^\circ.
$$

Das Produkt

$$
(2+3i)(4-i)
=
11+10i
$$

besitzt also den Winkel

$$
42^\circ.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

vectors = [
    (2, 3, r"$2+3i$"),
    (4, -1, r"$4-i$"),
    (11, 10, r"$11+10i$")
]

fig, ax = plt.subplots(figsize=(7, 7))

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

for x, y, label in vectors:
    ax.arrow(
        0, 0, x, y,
        length_includes_head=True,
        head_width=0.25
    )
    ax.text(x * 0.75, y * 0.75, label, fontsize=11)

ax.text(1.4, 0.8, r"$\theta_1\approx56^\circ$", fontsize=10)
ax.text(3.0, -0.8, r"$\theta_2\approx-14^\circ$", fontsize=10)
ax.text(6.0, 6.0, r"$\theta_1+\theta_2\approx42^\circ$", fontsize=10)

ax.set_xlim(-2, 12)
ax.set_ylim(-2, 11)
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Weiteres Beispiel der Multiplikation

Beispiel:

$$
(1+2i)(-1+i)
=
-3-i.
$$

Hier können beide Winkel als positiv angesehen werden:

| Zahl | Winkel |
|---|---:|
| $1+2i$ | $63.4^\circ$ |
| $-1+i$ | $135^\circ$ |
| $-3-i$ | $63.4^\circ+135^\circ=198.4^\circ$ |

Die Beträge werden multipliziert und die Winkel addiert, was zu einer Drehung und Skalierung führt.

```python
import matplotlib.pyplot as plt

vectors = [
    (1, 2, r"$1+2i$"),
    (-1, 1, r"$-1+i$"),
    (-3, -1, r"$-3-i$")
]

fig, ax = plt.subplots(figsize=(7, 7))

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

for x, y, label in vectors:
    ax.arrow(
        0, 0, x, y,
        length_includes_head=True,
        head_width=0.12
    )
    ax.text(x * 1.05, y * 1.05, label, fontsize=11)

ax.text(1.0, 0.6, r"$63.4^\circ$", fontsize=10)
ax.text(-0.6, 1.3, r"$135^\circ$", fontsize=10)
ax.text(-2.8, 0.7, r"$198.4^\circ$", fontsize=10)

ax.set_xlim(-4, 4)
ax.set_ylim(-3, 4)
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Division von komplexen Zahlen

**Division:**

- Multipliziere Zähler und Nenner mit der konjugiert komplexen Zahl des Nenners.
- Für

$$
z=a+bi
$$

ist die konjugierte Zahl

$$
\overline z=a-bi.
$$

**Divisionsformel:**

$$
\frac{a+bi}{c+di}
=
\frac{(a+bi)(c-di)}
     {(c+di)(c-di)}
$$

$$
=
\frac{(ac+bd)+(bc-ad)i}
     {c^2+d^2}.
$$

**Beispiel:**

$$
\frac{3+4i}{1-2i}
=
\frac{(3+4i)(1+2i)}
     {(1-2i)(1+2i)}
$$

$$
=
\frac{(3-8)+(4\cdot1+3\cdot2)i}
     {1^2+2^2}
$$

$$
=
\frac{-5+10i}{5}
=
-1+2i.
$$

### Graphische Darstellung der Division

Bei der Division zweier komplexer Zahlen wird

- der Betrag des Zählers durch den Betrag des Nenners geteilt,
- der Winkel des Nenners vom Winkel des Zählers subtrahiert.

Für

$$
\frac{3+4i}{1-2i}
$$

gilt:

$$
|3+4i|
=
\sqrt{3^2+4^2}
=
5
$$

und

$$
|1-2i|
=
\sqrt{1^2+2^2}
=
\sqrt5.
$$

Damit ist der Betrag des Ergebnisses

$$
|z|
=
\frac{5}{\sqrt5}
=
\sqrt5.
$$

Für die Winkel gilt näherungsweise:

$$
\theta_1
=
53.1^\circ,
$$

$$
\theta_2
=
-63.4^\circ.
$$

Daher

$$
\theta
=
\theta_1-\theta_2
=
53.1^\circ+63.4^\circ
=
116.5^\circ.
$$

Grafisch entspricht dies einer Skalierung und einer Drehung in entgegengesetzter Richtung zur Multiplikation.

```python
import matplotlib.pyplot as plt

vectors = [
    (3, 4, r"$3+4i$"),
    (1, -2, r"$1-2i$"),
    (-1, 2, r"$-1+2i$")
]

fig, ax = plt.subplots(figsize=(7, 7))

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

for x, y, label in vectors:
    ax.arrow(
        0, 0, x, y,
        length_includes_head=True,
        head_width=0.12
    )
    ax.text(x * 1.05, y * 1.05, label, fontsize=11)

ax.text(1.4, 0.7, r"$\theta_1=53.1^\circ$", fontsize=10)
ax.text(0.8, -1.0, r"$\theta_2=-63.4^\circ$", fontsize=10)
ax.text(-1.8, 2.4, r"$\theta=116.5^\circ$", fontsize=10)

ax.set_xlim(-2.5, 6)
ax.set_ylim(-3, 5)
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```

### Reziproke komplexe Zahlen: Beispiel

Eine komplexe Zahl und ihre reziproke Zahl (Kehrwert) können auf der Gauss-Ebene grafisch dargestellt werden.

Betrachte

$$
z
=
\frac{1}{\sqrt2}
+
\frac{1}{\sqrt2}i.
$$

Der Betrag von $z$ ist

$$
|z|
=
\sqrt{
\left(\frac{1}{\sqrt2}\right)^2
+
\left(\frac{1}{\sqrt2}\right)^2
}
=
1.
$$

Die reziproke Zahl

$$
\frac{1}{z}
$$

hat denselben Betrag, aber der Winkel (Argument) ist das Negative des Winkels von $z$.

```python
import numpy as np
import matplotlib.pyplot as plt

a = 1 / np.sqrt(2)

z = (a, a)
z_inv = (a, -a)

fig, ax = plt.subplots(figsize=(6, 6))

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

ax.arrow(
    0, 0, z[0], z[1],
    length_includes_head=True,
    head_width=0.05
)

ax.arrow(
    0, 0, z_inv[0], z_inv[1],
    length_includes_head=True,
    head_width=0.05
)

ax.text(0.45, 0.65, r"$z$", fontsize=12)
ax.text(0.45, -0.75, r"$\frac{1}{z}$", fontsize=12)

ax.text(0.45, 0.18, r"$\frac{\pi}{4}$", fontsize=11)
ax.text(0.45, -0.25, r"$-\frac{\pi}{4}$", fontsize=11)

ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_aspect("equal")
ax.grid(alpha=0.3)

plt.show()
```
