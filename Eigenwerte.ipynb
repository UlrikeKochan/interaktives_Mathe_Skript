# Eigenwerte und Eigenvektoren



<a id="masse-feder-system"></a>

## Masse-Feder-System

### Ein Masse-Feder-System

- Längere Schwingungsdauer (niedrigere Eigenfrequenz).
- Kürzere Schwingungsdauer (höhere Eigenfrequenz).

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, axes = plt.subplots(2, 1, figsize=(8, 5))

periods = [40, 20]
titles = [
    "Längere Schwingungsdauer (niedrigere Eigenfrequenz)",
    "Kürzere Schwingungsdauer (höhere Eigenfrequenz)"
]

points = []

for ax, title in zip(axes, titles):
    ax.set_xlim(-2, 2)
    ax.set_ylim(-0.6, 0.6)
    ax.axvline(0, linestyle="--")
    ax.set_yticks([])
    ax.set_title(title)

    point, = ax.plot([0], [0], "o", markersize=16)
    points.append(point)

def update(frame):
    for point, N in zip(points, periods):
        x = np.sin(2 * np.pi * frame / N)
        point.set_data([x], [0])
    return points

ani = FuncAnimation(
    fig,
    update,
    frames=80,
    interval=70,
    blit=True
)

plt.close(fig)
HTML(ani.to_jshtml())
```

### Freie Schwingung eines Masse-Feder-Systems

- In der Skizze ist ein Block mit Masse $m$ an einer horizontalen Feder mit Federkonstante $c$ befestigt.
- Zunächst wird die Masse aus der Ruhelage ausgelenkt oder kurz angestoßen (z.B. nach rechts gezogen und losgelassen).
- Danach wirkt keine äußere Kraft mehr:
  - Es wirkt nur die Rückstellkraft der Feder.
  - Das System führt eine **freie Schwingung** um die Ruhelage aus.

**Mathematisches Modell:**

$$
m\,\ddot{x}(t)+c\,x(t)=0
$$

Dabei ist

- $x(t)$: Auslenkung der Masse aus der Ruhelage,
- $m$: Masse des Blocks,
- $c$: Federsteifigkeit (Federkonstante).

Die gezeigte Schwingung ist idealisiert: keine Reibung, keine Dämpfung, die Amplitude bleibt zeitlich konstant.

<a id="eigenfrequenz"></a>

## Eigenfrequenz

Aus der Bewegungsgleichung

$$
m\,\ddot{x}(t)+c\,x(t)=0
$$

erhält man als Lösung

$$
x(t)=A\cos(\omega_0t+\varphi),
\qquad
\omega_0=\sqrt{\frac{c}{m}}.
$$

### Eigenfrequenz des Masse-Feder-Systems

- $\omega_0$ heißt **Eigenkreisfrequenz** der Schwingung.
- Für die Schwingungsdauer $T$ und die Frequenz $f$ gilt:

$$
T=\frac{2\pi}{\omega_0}
=
2\pi\sqrt{\frac{m}{c}},
\qquad
f=\frac{1}{T}
=
\frac{\omega_0}{2\pi}.
$$

### Eigenfrequenz – Interpretation

**Größere Federsteifigkeit $c$:**

- Die Feder ist „härter“, die Rückstellkraft ist größer.
- Die Masse schwingt schneller um die Ruhelage.
- Die Eigenfrequenz steigt.
- Die Schwingungsdauer $T$ wird kleiner.

**Größere Masse $m$:**

- Das System ist träger.
- Die Schwingung wird langsamer.
- Die Periode $T$ wird größer.
- Die Eigenfrequenz nimmt ab.

### Massen-Feder-System mit zwei Massen

**Hookesches Gesetz:**

$$
F=-c\,\Delta x
$$

Dabei ist

- $F$: Rückstellkraft der Feder $([\mathrm{N}])$,
- $c$: Federkonstante $([\mathrm{N/m}])$,
- $\Delta x$: Auslenkung der Feder aus ihrer Ruhelage $([\mathrm{m}])$.

<a id="normalschwingungen"></a>

## Normalschwingungen

**Anwendung auf das Massen-Feder-System:**

Man betrachte ein System aus zwei identischen Massenpunkten

$$
m_1=m_2=m,
$$

die durch eine Feder mit der Konstante $c$ verbunden sind.

Die Koordinaten $x_1$ und $x_2$ beschreiben die momentane Lage der Massen und erfüllen idealisiert die Differentialgleichungen

$$
\begin{aligned}
m\ddot{x}_1 &= -c(x_1-x_2),\\
m\ddot{x}_2 &= -c(x_2-x_1).
\end{aligned}
$$

**Lösungsansatz:**

Der Ansatz

$$
x_k=A_ke^{i\omega t}
$$

führt zu

$$
\ddot{\vec{x}}
=
-\omega^2\vec{x}.
$$

Hierbei bestimmt die Federkonstante $c$ zusammen mit $m$ die Winkelgeschwindigkeit $\omega$ der Schwingungen.

### Normalschwingungen: Eigenwertproblem

Das Gleichungssystem wird zu

$$
\underbrace{
\begin{pmatrix}
-\frac{c}{m} & \frac{c}{m}\\
\frac{c}{m} & -\frac{c}{m}
\end{pmatrix}
}_{\mathbf A}
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
=
-\omega^2
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}.
$$

Die Schwingungsdauer ist für $\omega>0$

$$
T=\frac{2\pi}{\omega}
\qquad [\mathrm{s}].
$$

Durch die Substitution

$$
\lambda=-\omega^2
$$

wird das Problem zu einem Eigenwertproblem:

$$
\mathbf A\vec{x}
=
\lambda\vec{x}.
$$

```python
# Ergänzung

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def normale_schwingung(modus=1, phase=0.0):
    t = phase

    if modus == 1:
        x1 = np.sin(t)
        x2 = -np.sin(t)
        titel = "Antisymmetrische Schwingung"
    else:
        x1 = np.sin(t)
        x2 = np.sin(t)
        titel = "Symmetrische Translation"

    fig, ax = plt.subplots(figsize=(8, 2.5))

    ax.axvline(-1.5, linestyle="--")
    ax.axvline(1.5, linestyle="--")

    ax.plot([-1.5 + x1, 1.5 + x2], [0, 0], linewidth=2)
    ax.scatter([-1.5 + x1, 1.5 + x2], [0, 0], s=500)

    ax.set_xlim(-4, 4)
    ax.set_ylim(-1, 1)
    ax.set_yticks([])
    ax.set_title(titel)

    plt.show()

widgets.interact(
    normale_schwingung,
    modus=widgets.ToggleButtons(
        options=[("Eigenmode 1", 1), ("Eigenmode 2", 2)]
    ),
    phase=widgets.FloatSlider(
        min=0,
        max=2*np.pi,
        step=0.1,
        value=0
    )
)
```

## Eigenwerte und Eigenvektoren

Wir betrachten die Beziehung zwischen der Matrix $\mathbf A$ und dem Vektor $\mathbf v_1$.

Die Transformation des Vektors durch die Matrix ändert lediglich seine Länge, nicht aber seine Richtung:

$$
\mathbf A
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix},
\qquad
\mathbf v_1
=
\begin{pmatrix}
1\\
1
\end{pmatrix}.
$$

Es gilt:

$$
\mathbf A\mathbf v_1
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix}
\begin{pmatrix}
1\\
1
\end{pmatrix}
=
\begin{pmatrix}
2\\
2
\end{pmatrix}
=
2\mathbf v_1.
$$

Es gilt die Beziehung

$$
\mathbf A\mathbf v_1
=
\lambda\mathbf v_1.
$$

Dies bedeutet, dass der Vektor $\mathbf v_1$ durch die Matrix $\mathbf A$ um den Faktor

$$
\lambda=2
$$

skaliert wird.

Ist dies der einzige Vektor mit dieser Eigenschaft?

Ein weiterer Vektor mit derselben Eigenschaft, aber einem anderen Skalierungsfaktor $\lambda$, ist

$$
\mathbf v_2
=
\begin{pmatrix}
1\\
2
\end{pmatrix}.
$$

Es gilt

$$
\mathbf A\mathbf v_2
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix}
\begin{pmatrix}
1\\
2
\end{pmatrix}
=
\begin{pmatrix}
-2\\
-4
\end{pmatrix}
=
-2\mathbf v_2.
$$

Die Vektoren $\mathbf v_1$ und $\mathbf v_2$, die diese Eigenschaft haben, heißen **Eigenvektoren** der Matrix $\mathbf A$, und die Skalierungsfaktoren $\lambda_1$ und $\lambda_2$ heißen **Eigenwerte**.

### Eigenwertproblem

Gesucht sind alle Paare $(\lambda,\mathbf v)$, für die gilt:

$$
\mathbf A\mathbf v
=
\lambda\mathbf v,
$$

mit

$$
\mathbf A\in\mathbb K^{n\times n},
\qquad
\lambda\in\mathbb K,
\qquad
\mathbf v\in\mathbb K^n.
$$

<a id="eigenschaften-eigenvektoren"></a>

## Eigenschaften von Eigenvektoren

**Wichtig:**

Falls $(\lambda,\mathbf v)$ das Eigenwertproblem löst, dann löst auch

$$
(\lambda,c\mathbf v)
$$

für beliebige $c\in\mathbb K$ das Problem:

$$
\mathbf A(c\mathbf v)
=
c\mathbf A\mathbf v
=
c(\lambda\mathbf v)
=
\lambda(c\mathbf v).
$$

Dies bedeutet, dass nur die Richtung der Eigenvektoren relevant ist.

Sie können beliebig skaliert werden und behalten trotzdem ihre Eigenschaft als Eigenvektoren bei.

### Existenz von Eigenvektoren: Ein Gegenbeispiel

Besitzt jede $n\times n$-Matrix genau $n$ Eigenvektoren?

Die Antwort ist **nein**.

Betrachten wir

$$
\mathbf A
=
\begin{bmatrix}
1 & -2\\
0 & 1
\end{bmatrix}.
$$

Wir suchen $\lambda\in\mathbb K$ und $\mathbf x\in\mathbb K^2$, sodass

$$
\mathbf A
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
=
\lambda
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}.
$$

Das Beispiel zeigt: Die Anzahl der Eigenvektoren hängt von der Struktur der Matrix ab.

### Eigenwertproblem mit nur einem Eigenvektor

Aus den Gleichungen

$$
x_1-2x_2=\lambda x_1,
\qquad
x_2=\lambda x_2
$$

ergibt sich

$$
\lambda=1.
$$

Für $\lambda=1$ folgt aus der ersten Gleichung:

$$
x_1-2x_2=x_1
\quad\Rightarrow\quad
x_2=0.
$$

Der Eigenvektor hat die Form

$$
\mathbf v
=
\begin{pmatrix}
x_1\\
0
\end{pmatrix},
\qquad
x_1\in\mathbb K.
$$

Dies zeigt, dass die Matrix nur einen Eigenvektor besitzt.

Als Repräsentant kann beispielsweise

$$
\mathbf v
=
\begin{pmatrix}
1\\
0
\end{pmatrix}
$$

gewählt werden.

Die Matrix $\mathbf A$ besitzt:

- nur einen Eigenwert: $\lambda=1$,
- nur einen zugehörigen Eigenvektor:

$$
\mathbf v
=
\begin{pmatrix}
1\\
0
\end{pmatrix}.
$$

Dieses Beispiel illustriert, dass die Anzahl der Eigenvektoren einer Matrix nicht immer ihrer Dimension entspricht.

<a id="loesung-eigenwertproblem"></a>

## Lösung des Eigenwertproblems

Die Lösung des Eigenwertproblems erfolgt in zwei Schritten:

1. Bestimmung der Eigenwerte: Diese sind die Nullstellen des sogenannten charakteristischen Polynoms.
2. Bestimmung der zugehörigen Eigenvektoren für die gefundenen Eigenwerte.

Das charakteristische Polynom ergibt sich als Determinante der Matrix

$$
\mathbf A-\lambda\mathbf I.
$$

Gegeben sei

$$
\mathbf A
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix}.
$$

Die Matrix $\mathbf A_\lambda$ lautet

$$
\mathbf A_\lambda
=
\begin{bmatrix}
6-\lambda & -4\\
8 & -6-\lambda
\end{bmatrix}.
$$

<a id="charakteristisches-polynom"></a>

## Das charakteristische Polynom

Die Matrix zur Bestimmung des charakteristischen Polynoms lautet

$$
\mathbf A_\lambda
=
\mathbf A-\lambda\mathbf I.
$$

Also

$$
\mathbf A_\lambda
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix}
-
\lambda
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}.
$$

### Charakteristisches Polynom

Das charakteristische Polynom wird definiert als

$$
p(\lambda)
:=
\det(\mathbf A_\lambda).
$$

### Berechnung der Eigenwerte

Zur Bestimmung der Eigenwerte berechnen wir die Determinante von $\mathbf A_\lambda$:

$$
\det(\mathbf A_\lambda)
=
\begin{vmatrix}
6-\lambda & -4\\
8 & -6-\lambda
\end{vmatrix}.
$$

Damit

$$
\det(\mathbf A_\lambda)
=
(6-\lambda)(-6-\lambda)-8(-4)
$$

und somit

$$
\det(\mathbf A_\lambda)
=
\lambda^2-4.
$$

Faktorisieren ergibt

$$
\lambda^2-4
=
(\lambda-2)(\lambda+2).
$$

Die Eigenwerte ergeben sich als Nullstellen des charakteristischen Polynoms:

$$
\boxed{\lambda_1=2,\qquad\lambda_2=-2}.
$$

```python
# Ergänzung

import sympy as sp

A = sp.Matrix([
    [6, -4],
    [8, -6]
])

lam = sp.symbols("lambda")

p = sp.factor((A - lam * sp.eye(2)).det())

print("Charakteristisches Polynom:")
sp.pprint(p)

print("\nEigenwerte und algebraische Vielfachheiten:")
print(A.eigenvals())
```

### Eigenwert mit doppelter Vielfachheit: Ein Beispiel

Betrachten wir die Matrix

$$
\mathbf A
=
\begin{bmatrix}
-2 & 1\\
-1 & 0
\end{bmatrix}.
$$

Das charakteristische Polynom berechnet sich aus

$$
\det(\mathbf A_\lambda)
=
\begin{vmatrix}
-2-\lambda & 1\\
-1 & -\lambda
\end{vmatrix}.
$$

Damit

$$
=(-2-\lambda)(-\lambda)-(-1)\cdot1
$$

und

$$
=\lambda^2+2\lambda+1.
$$

Also

$$
p(\lambda)
=
(\lambda+1)^2.
$$

Das charakteristische Polynom hat eine doppelte Nullstelle bei

$$
\lambda=-1.
$$

<a id="algebraische-vielfachkeit"></a>

## Algebraische Dimension / Vielfachkeit von Eigenwerten

Die Nullstellen des charakteristischen Polynoms sind nicht immer einfach.

Sie können eine Vielfachheit größer als $1$ aufweisen.

### Algebraische Dimension / Algebraische Vielfachheit

Die **algebraische Dimension** eines Eigenwerts $\lambda_j$ wird durch die Vielfachheit $k_j$ seiner Nullstelle im charakteristischen Polynom definiert.

Wichtige Eigenschaften:

- Das charakteristische Polynom hat über $\mathbb C$ immer genau $n$ Nullstellen.
- Die Summe der algebraischen Dimensionen aller Eigenwerte ergibt die Matrixdimension:

$$
\sum_j k_j=n.
$$

<a id="bestimmung-eigenvektoren"></a>

## Schritte zur Bestimmung der Eigenvektoren

Gegeben:

$$
\mathbf A
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix},
\qquad
\lambda_1=2,
\qquad
\lambda_2=-2.
$$

Die Eigenvektoren erfüllen

$$
\mathbf A\mathbf v_1
=
\lambda_1\mathbf v_1,
\qquad
\mathbf A\mathbf v_2
=
\lambda_2\mathbf v_2.
$$

Für jeden Eigenwert $\lambda_i$ bestimmen wir die zugehörigen Eigenvektoren $\mathbf v_i$ durch Lösung des Gleichungssystems.

### Berechnung des Eigenvektors $\mathbf v_1$ für $\lambda_1=2$

Wir lösen

$$
\mathbf A\mathbf v_1
=
2\mathbf v_1.
$$

Dies kann umgeschrieben werden als

$$
\mathbf A\mathbf v_1
-
2\mathbf I\mathbf v_1
=
\mathbf 0,
$$

also

$$
(\mathbf A-2\mathbf I)\mathbf v_1
=
\mathbf 0.
$$

Dabei ist

$$
\mathbf I
=
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}.
$$

Damit gilt

$$
\mathbf v_1
\in
\operatorname{Kern}(\mathbf A-2\mathbf I).
$$

**Schritte zur Bestimmung von Eigenvektoren:**

1. Bestimme den Kern der Matrix

$$
\mathbf A-\lambda\mathbf I.
$$

2. Verwende das Verfahren der Gauß-Elimination.

Das Eigenwertproblem reduziert sich somit auf die Berechnung des Kerns der modifizierten Matrix.

### Gauß-Elimination zur Bestimmung des Kerns

Für $\lambda_1=2$ gilt

$$
\mathbf A-2\mathbf I
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix}
-
\begin{bmatrix}
2 & 0\\
0 & 2
\end{bmatrix}
=
\begin{bmatrix}
4 & -4\\
8 & -8
\end{bmatrix}.
$$

Gauß-Elimination:

$$
\begin{bmatrix}
4 & -4\\
8 & -8
\end{bmatrix}
\quad
\xrightarrow{\mathrm{II'}=\mathrm{II}-2\mathrm I}
\quad
\begin{bmatrix}
4 & -4\\
0 & 0
\end{bmatrix}.
$$

Wir lösen das homogene Gleichungssystem

$$
\begin{bmatrix}
4 & -4\\
0 & 0
\end{bmatrix}
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
=
\begin{pmatrix}
0\\
0
\end{pmatrix}.
$$

Aus der ersten Gleichung folgt

$$
x_1=x_2=t,
\qquad
t\in\mathbb R.
$$

Daraus ergibt sich der Eigenvektor

$$
\mathbf v_1
=
t
\begin{pmatrix}
1\\
1
\end{pmatrix},
\qquad
t\in\mathbb R.
$$

**Basis des Kerns:**

$$
B
=
\left\{
\begin{pmatrix}
1\\
1
\end{pmatrix}
\right\}.
$$

Damit ist die Richtung des Eigenvektors eindeutig bestimmt.

<a id="geometrische-vielfachkeit"></a>

## Geometrische Dimension / Vielfachheit eines Eigenwerts

Die Dimension des Kerns der Matrix

$$
\mathbf A-\lambda_j\mathbf I
$$

wird als **geometrische Dimension** des Eigenwertes $\lambda_j$ bezeichnet und kann auch als **geometrische Vielfachkeit** des Eigenwerts bezeichnet werden.

$$
\rho(\lambda_j)
=
\dim\left(
\operatorname{Kern}
(\mathbf A-\lambda_j\mathbf I)
\right).
$$

Die geometrische Dimension oder geometrische Vielfachkeit gibt die Anzahl linear unabhängiger Eigenvektoren an, die zu einem Eigenwert $\lambda_j$ gehören.

### Berechnung des Eigenvektors $\mathbf v_2$ für $\lambda_2=-2$

Wir berechnen

$$
\mathbf A-\lambda_2\mathbf I
=
\mathbf A+2\mathbf I.
$$

Damit

$$
\mathbf A+2\mathbf I
=
\begin{bmatrix}
6 & -4\\
8 & -6
\end{bmatrix}
+
\begin{bmatrix}
2 & 0\\
0 & 2
\end{bmatrix}
=
\begin{bmatrix}
8 & -4\\
8 & -4
\end{bmatrix}.
$$

Mit

$$
\mathrm{II'}=\mathrm{II}-\mathrm I
$$

folgt die reduzierte Matrix

$$
\begin{bmatrix}
8 & -4\\
0 & 0
\end{bmatrix}.
$$

Wir lösen

$$
\begin{bmatrix}
8 & -4\\
0 & 0
\end{bmatrix}
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
=
\begin{pmatrix}
0\\
0
\end{pmatrix}.
$$

Es folgt

$$
x_1=\frac12t,
\qquad
x_2=t,
\qquad
t\in\mathbb R.
$$

Die Lösungsmenge ist

$$
L
:=
\left\{
\mathbf x\in\mathbb R^2
\;\middle|\;
\mathbf x
=
t
\begin{pmatrix}
\frac12\\
1
\end{pmatrix},
\;
t\in\mathbb R
\right\}.
$$

Der Eigenvektor hat somit die Richtung

$$
\mathbf v_2
=
t
\begin{pmatrix}
\frac12\\
1
\end{pmatrix}.
$$

### Basis und normierte Eigenvektoren der Matrix $\mathbf A$

**Basis des Kerns von $(\mathbf A+2\mathbf I)$:**

$$
B
=
\left\{
\begin{pmatrix}
\frac12\\
1
\end{pmatrix}
\right\}.
$$

**Eigenvektor $\mathbf v_2$:**

$$
\mathbf v_2
=
\begin{pmatrix}
\frac12\\
1
\end{pmatrix}.
$$

### Zusammenfassung der Eigenwerte und Eigenvektoren

$$
\lambda_1=2,
\qquad
\mathbf v_1=
\begin{pmatrix}
1\\
1
\end{pmatrix},
$$

$$
\lambda_2=-2,
\qquad
\mathbf v_2=
\begin{pmatrix}
\frac12\\
1
\end{pmatrix}.
$$

**Normierte Eigenvektoren:**

$$
\tilde{\mathbf v}_1
=
\frac{1}{\sqrt2}
\begin{pmatrix}
1\\
1
\end{pmatrix},
$$

$$
\tilde{\mathbf v}_2
=
\frac{2}{\sqrt5}
\begin{pmatrix}
\frac12\\
1
\end{pmatrix}.
$$

Die Normierung bringt die Eigenvektoren auf die Länge $1$, was in vielen Anwendungen hilfreich ist.

```python
# Ergänzung

import sympy as sp

A = sp.Matrix([
    [6, -4],
    [8, -6]
])

for eigenwert, vielfachheit, basis in A.eigenvects():
    print("Eigenwert:", eigenwert)
    print("Algebraische Vielfachheit:", vielfachheit)
    print("Basis des Eigenraums:")

    for v in basis:
        sp.pprint(v)

    print()
```

### Geometrische und algebraische Dimension / Vielfachheit

**Gegeben:**

$$
\mathbf A\in\mathbb K^{n\times n},
$$

mit Eigenwerten $\lambda_j$.

- **Geometrische Dimension / Vielfachheit $\rho(\lambda_j)$:**  
  Anzahl linear unabhängiger Eigenvektoren, die zu $\lambda_j$ gehören.

- **Algebraische Dimension / Vielfachheit $k_j$:**  
  Vielfachheit des Eigenwerts $\lambda_j$ als Nullstelle des charakteristischen Polynoms.

Es gilt:

$$
\rho(\lambda_j)
\leq
k_j
\qquad
\forall j.
$$

Wann gilt

$$
\rho(\lambda_j)=k_j
\qquad
\forall j?
$$

Geometrische Dimension entspricht algebraischer Dimension genau dann, wenn

$$
n
=
\sum_j\rho(\lambda_j).
$$

**Bedeutung:**

In diesem Fall existiert in $\mathbb K^n$ eine Basis, die vollständig aus Eigenvektoren von $\mathbf A$ besteht.

<a id="diagonalisierbarkeit"></a>

## Definition und Kriterium einer diagonalisierbaren Matrix

Eine quadratische Matrix

$$
\mathbf A\in\mathbb K^{n\times n}
$$

heißt **diagonalisierbar**, wenn es eine invertierbare Matrix

$$
\mathbf P\in\mathbb K^{n\times n}
$$

gibt, sodass

$$
\mathbf P^{-1}\mathbf A\mathbf P
=
\mathbf D,
$$

wobei $\mathbf D$ eine Diagonalmatrix ist.

**Eigenschaften:**

- Die Diagonalelemente von $\mathbf D$ sind die Eigenwerte der Matrix $\mathbf A$.
- Die Spalten von $\mathbf P$ sind die zugehörigen Eigenvektoren von $\mathbf A$.

**Kriterium für Diagonalisierbarkeit:**

Eine Matrix $\mathbf A$ ist genau dann diagonalisierbar, wenn

$$
\rho(\lambda_j)
=
k_j
\qquad
\forall j.
$$

Das bedeutet, dass die Matrix $\mathbf A$ eine Basis aus Eigenvektoren besitzt.

**Anwendung:**

Diagonalisierbare Matrizen vereinfachen Berechnungen wie Potenzen und Exponentialen der Matrix.

<a id="positive-definitheit"></a>

## Positive Definitheit und Eigenwerte

Eine quadratische Matrix

$$
\mathbf A\in\mathbb C^{n\times n}
$$

wird als **positiv definit** bezeichnet, wenn für alle Nicht-Null-Vektoren

$$
\mathbf x\in\mathbb C^n
$$

gilt:

$$
\mathbf x^*\mathbf A\mathbf x>0,
$$

wobei $\mathbf x^*$ das konjugiert-transponierte von $\mathbf x$ ist.

Eine wichtige Eigenschaft von positiv definiten Matrizen ist, dass alle Eigenwerte von $\mathbf A$ positiv und reell sind.

### Eigenwerte und positive Definitheit

- $\lambda>0$ für alle Eigenwerte von $\mathbf A$ $\implies$ $\mathbf A$ ist positiv definit.
- Gibt es Eigenwerte mit $\lambda\leq0$ oder $\lambda\notin\mathbb R$, dann ist $\mathbf A$ nicht positiv definit.

### Beispiel: Überprüfung der positiven Definitheit

Gegeben sei

$$
\mathbf A
=
\begin{bmatrix}
2 & -1\\
-1 & 2
\end{bmatrix}.
$$

Wir bestimmen die Eigenwerte:

$$
\mathbf A_\lambda
=
\begin{bmatrix}
2-\lambda & -1\\
-1 & 2-\lambda
\end{bmatrix}.
$$

Damit

$$
\det(\mathbf A_\lambda)
=
(2-\lambda)^2-(-1)^2
=
\lambda^2-4\lambda+3.
$$

Faktorisieren ergibt

$$
\lambda^2-4\lambda+3
=
(\lambda-3)(\lambda-1).
$$

Die Eigenwerte sind

$$
\lambda_1=3,
\qquad
\lambda_2=1.
$$

Da alle Eigenwerte positiv sind, ist die Matrix $\mathbf A$ **positiv definit**.

### Beispiel: Hermitesche und positiv definite Matrix

Betrachten wir

$$
\mathbf A
=
\begin{pmatrix}
2 & 1+i\\
1-i & 3
\end{pmatrix}.
$$

**Hermitesch:**

Die Matrix erfüllt

$$
\mathbf A=\mathbf A^*,
$$

da die Diagonaleinträge reell sind und die Nebendiagonaleinträge konjugiert zueinander sind:

$$
\mathbf A^*
=
\begin{pmatrix}
2 & 1+i\\
1-i & 3
\end{pmatrix}.
$$

**Eigenwerte:**

Die Eigenwerte werden aus

$$
\det(\mathbf A-\lambda\mathbf I)=0
$$

berechnet:

$$
\det
\begin{pmatrix}
2-\lambda & 1+i\\
1-i & 3-\lambda
\end{pmatrix}
=
\lambda^2-5\lambda+4
=
0.
$$

Die Lösungen sind

$$
\lambda=4
\qquad\text{und}\qquad
\lambda=1.
$$

Da alle Eigenwerte positiv sind und die Matrix hermitesch ist, ist $\mathbf A$ positiv definit.

### Zusammenhang mit quadratischen Formen

Eine positiv definite Matrix

$$
\mathbf A\in\mathbb C^{n\times n}
$$

hat die Eigenschaft, dass die zugehörige quadratische Form immer positiv ist:

$$
Q(\mathbf x)
=
\mathbf x^*\mathbf A\mathbf x
>
0
\qquad
\forall\mathbf x\in\mathbb C^n,
\quad
\mathbf x\neq\mathbf 0.
$$

Dabei ist $\mathbf x^*$ das konjugiert-transponierte von $\mathbf x$.

**Eigenschaften:**

- Positive Eigenwerte der hermiteschen Matrix $\mathbf A$ implizieren

$$
Q(\mathbf x)>0.
$$

- Die Matrix $\mathbf A$ muss hermitesch sein,

$$
\mathbf A=\mathbf A^*,
$$

damit die quadratische Form reell ist.

- Die quadratische Form $Q(\mathbf x)$ beschreibt Energie, Stabilität oder Varianz in vielen physikalischen und statistischen Anwendungen.

### Implikationen positiver Eigenwerte für die Energiedefinition

- **Positive Definitheit:**  
  Positive Eigenwerte einer Energie-Matrix oder eines Operators bedeuten, dass die Energie nach unten beschränkt ist, z.B.

  $$
  E\geq0.
  $$

  Dies gewährleistet die Stabilität des Systems.

- **Physikalische Validität:**  
  Positive Eigenwerte stellen sicher, dass die Energie physikalisch sinnvoll, d.h. nicht negativ, ist.

- **Energiezustände:**  
  Positive Eigenwerte stehen für gut definierte Energiezustände, z.B. in der Quantenmechanik oder bei Schwingungsmoden.

## Anwendungen von Eigenwerten und Eigenvektoren

Eigenwerte und Eigenvektoren finden breite Anwendung in verschiedenen Disziplinen der Mathematik, Physik und Datenwissenschaft:

- **Hauptkomponentenanalyse (PCA):**  
  Reduktion der Dimension von Datensätzen durch Finden der Hauptrichtungen maximaler Varianz.

- **Stabilitätsanalyse:**

## Massen-Feder-System

**Hookesches Gesetz:**

$$
F=-c\,\Delta x.
$$

### Normalschwingungen

Differentialgleichungen:

$$
\begin{aligned}
m\ddot{x}_1 &= -c(x_1-x_2),\\
m\ddot{x}_2 &= -c(x_2-x_1).
\end{aligned}
$$

**Lösungsansatz:**

Der Ansatz

$$
x_k=A_ke^{i\omega t}
$$

führt zu

$$
\ddot{\vec{x}}
=
-\omega^2\vec{x}.
$$

Frequenz:

$$
\omega^2=\frac{c}{m}.
$$

### Normalschwingungen: Eigenwertproblem

Das Gleichungssystem lautet

$$
\underbrace{
\begin{pmatrix}
-\frac{c}{m} & \frac{c}{m}\\
\frac{c}{m} & -\frac{c}{m}
\end{pmatrix}
}_{\mathbf A}
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}
=
-\omega^2
\begin{pmatrix}
x_1\\
x_2
\end{pmatrix}.
$$

Durch Substitution

$$
\lambda=-\omega^2
$$

wird das Problem zu einem Eigenwertproblem:

$$
\mathbf A\vec{x}
=
\lambda\vec{x}.
$$

### Aufgabenstellung

Man betrachtet das Eigenwertproblem für

$$
c=1\, [\mathrm{N/m}],
\qquad
m=250\, [\mathrm{g}].
$$

**Aufgaben:**

1. Schreibe die Matrix des Systems im internationalen Einheitensystem.
2. Bestimme die Eigenwerte $\lambda_1\,[\mathrm{s}^{-2}]$ und $\lambda_2\,[\mathrm{s}^{-2}]$.
3. Bestimme die Eigenvektoren.
4. Berechne die Schwingungsdauer der beiden Eigenschwingungen $T_1$ und $T_2$.

### Lösung

**1. Systemmatrix**

Da

$$
m=250\,\mathrm{g}=0.25\,\mathrm{kg},
$$

gilt

$$
\mathbf A
=
\begin{pmatrix}
-4 & 4\\
4 & -4
\end{pmatrix}.
$$

**2. Eigenwerte**

$$
\lambda_1=-8,
\qquad
\lambda_2=0.
$$

**3. Eigenvektoren**

$$
\vec{x}_1
=
\begin{pmatrix}
1\\
-1
\end{pmatrix},
\qquad
\vec{x}_2
=
\begin{pmatrix}
1\\
1
\end{pmatrix}.
$$

**4. Schwingungsdauer**

$$
\omega_1=2\sqrt2,
\qquad
\omega_2=0.
$$

Damit

$$
T_1
=
\frac{2\pi}{\omega_1}
=
\frac{\pi}{\sqrt2}
\,[\mathrm{s}].
$$

Für $T_2$ ist keine Schwingungsdauer definiert, da

$$
\omega_2=0.
$$

```python
# Ergänzung

import sympy as sp

c = sp.Rational(1, 1)
m = sp.Rational(1, 4)

A = sp.Matrix([
    [-c/m,  c/m],
    [ c/m, -c/m]
])

print("Systemmatrix A:")
sp.pprint(A)

print("\nEigenwerte und Eigenvektoren:")

for eigenwert, vielfachheit, basis in A.eigenvects():
    print("\nEigenwert:")
    sp.pprint(eigenwert)

    print("Eigenvektor:")
    for v in basis:
        sp.pprint(v)

    if eigenwert < 0:
        omega = sp.sqrt(-eigenwert)
        T = sp.simplify(2 * sp.pi / omega)

        print("omega =")
        sp.pprint(omega)

        print("T =")
        sp.pprint(T)

    elif eigenwert == 0:
        print("omega = 0 -> keine Schwingungsdauer definiert")
```

### Physikalische Interpretation des Eigenvektors $\vec{x}_1$

**Eigenvektor:**

$$
\vec{x}_1
=
\begin{pmatrix}
1\\
-1
\end{pmatrix}.
$$

**Interpretation: Antisymmetrische Schwingung**

- Die beiden Massen schwingen in entgegengesetzten Richtungen mit der gleichen Amplitude.

**Relative Bewegung:**

- Es kommt zur Deformation der Feder, da die beiden Massen relativ zueinander schwingen.

**Frequenz:**

- Dieser Eigenvektor ist mit dem Eigenwert

$$
\lambda_1=-8
$$

verknüpft.

- Die Winkelgeschwindigkeit beträgt

$$
\omega_1=2\sqrt2,
$$

was eine schnelle Schwingung aufgrund der Federsteifigkeit und Kopplung der Massen bedeutet.

### Physikalische Interpretation des Eigenvektors $\vec{x}_2$

**Eigenvektor:**

$$
\vec{x}_2
=
\begin{pmatrix}
1\\
1
\end{pmatrix}.
$$

**Interpretation: Symmetrische Translation**

- Die beiden Massen bewegen sich in die gleiche Richtung mit der gleichen Amplitude.
- Dies entspricht einer **starren Translation** des gesamten Systems.

**Relative Bewegung:**

- Es gibt keine Relativbewegung zwischen den Massen.
- Die Feder bleibt unverändert, ohne Stauchung oder Dehnung.

**Frequenz:**

- Dieser Eigenvektor ist mit dem Eigenwert

$$
\lambda_2=0
$$

verknüpft.

- Die Winkelgeschwindigkeit beträgt

$$
\omega_2=0,
$$

was keine Schwingung impliziert. Der Modus beschreibt nur eine statische Verschiebung.
